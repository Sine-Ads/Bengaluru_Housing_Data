```python
import pandas as pd
df = pd.read_csv('../data/Bengaluru_House_Data.csv')
df.head()
```

```python
def safe_float(x):
    try:
        return float(x)
    except (ValueError, TypeError):
        return None  

df[df['total_sqft'].apply(safe_float).isna()]
```

```python
problem_rows = df[df['total_sqft'].apply(safe_float).isna()]
problem_rows['total_sqft'].unique()
problem_rows.value_counts()
```

```python

def sqft_to_avg(x):
    parts = str(x).split('-')
    if len(parts) == 2:
        return (float(parts[0]) + float(parts[1])) / 2
    try:
        return float(x)
    except (ValueError, TypeError):
        return None

# Convert the whole column in-place
df['total_sqft'] = df['total_sqft'].apply(sqft_to_avg)
```

```python
def size_to_no(x):
    if pd.isna(x):
        return None
    try:
        # split() without arguments handles single or multiple spaces automatically
        return int(str(x).split()[0])
    except (ValueError, TypeError, IndexError):
        return None

# Apply to dataframe
df['size'] = df['size'].apply(size_to_no)
```
